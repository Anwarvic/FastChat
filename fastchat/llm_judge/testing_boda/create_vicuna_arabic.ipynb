```python
from datasets import load_dataset
import json

# Load dataset
ds = load_dataset("FreedomIntelligence/Arabic-Vicuna-80", split="train")

output_file = "/home/abdelrahman.sadallah/mbzuai/FastChat/fastchat/llm_judge/data/vicuna_bench_arabic/arabic_vicuna80_converted.jsonl"

with open(output_file, "w", encoding="utf-8") as outf:
    for example in ds:
        new_item = {
            "question_id": example["id"],
            "category": example["label"],
            "turns": [example["query"]]
        }
        outf.write(json.dumps(new_item, ensure_ascii=False) + "\n")

print(f"Converted dataset saved to {output_file}")
```
